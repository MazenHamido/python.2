```python
def bigger_than_seven(words,number):

    result =[]
    for word in words:
        if len(word) > number:
            result.append(word)

    return result


words = ["apple", "pie", "banana"]
number = 3

print(bigger_than_seven(words, number))
```

```python
def area(shape, *dimensions):
    if shape == "circle":
        radius = dimensions[0]
        result = 3.14 * radius * radius
        print("Circle Area:", result)

    elif shape == "rectangle":
        
        length = dimensions[0]
        width = dimensions[1]
        result = length * width
        print("Rectangle Area:", result)

    else:
        print("Invalid shape")


area("circle", 5)
area("rectangle", 4, 6)
```

```python
class Restaurant:
    def __init__(self):
        self.menu_items = {}
        self.book_table = []
        self.customer_orders = []

    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price

    def book_tables(self, customer, table_number):
        self.book_table.append({
            "customer": customer,
            "table": table_number
        })

    def customer_order(self, customer, item):
        self.customer_orders.append({
            "customer": customer,
            "item": item
        })

    def print_menu(self):
        print("Menu:", self.menu_items)

    def print_reservations(self):
        print("Reservations:", self.book_table)

    def print_orders(self):
        print("Customer Orders:", self.customer_orders)


restaurant = Restaurant()

restaurant.add_item_to_menu("Pizza", 150)
restaurant.add_item_to_menu("Burger", 100)
restaurant.add_item_to_menu("Pasta", 120)

restaurant.book_tables("Ahmed", 1)
restaurant.book_tables("Ali", 2)

restaurant.customer_order("Ahmed", "Pizza")
restaurant.customer_order("Ali", "Burger")

restaurant.print_menu()
restaurant.print_reservations()
restaurant.print_orders()
```

```python

class Customer:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    def get_age(self):
        return self.age

    def set_age(self, age):
        self.age = age


class Bike:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    def get_price(self):
        return self.price

    def set_price(self, price):
        self.price = price


class MainBikeRental:
    def __init__(self):
        self.bikes = []
        self.rentals = []

    def get_bikes(self):
        return self.bikes

    def set_bikes(self, bikes):
        self.bikes = bikes

    def get_rentals(self):
        return self.rentals

    def set_rentals(self, rentals):
        self.rentals = rentals

    def addBike(self, bike):
        self.bikes.append(bike)

    def requestBike(self, customer, number_of_bikes):
        if customer.get_age() <= 6:
            print("Customer must be older than 6")
            return

        if number_of_bikes <= 0:
            print("Invalid number of bikes")
            return

        if len(self.bikes) >= number_of_bikes:
            rented_bikes = self.bikes[:number_of_bikes]
            self.bikes = self.bikes[number_of_bikes:]

            self.rentals.append(rented_bikes)

            print("Bikes rented successfully")
        else:
            print("Wait 10 mins")

    def returnBike(self, rentalTime, num_of_bikes):
        if not self.rentals:
            print("No bikes have been rented")
            return

        rented_bikes = self.rentals[-1]

        if num_of_bikes != len(rented_bikes):
            print("Return the correct number of bikes")
            return

        bill = rentalTime * 40

        self.bikes.extend(rented_bikes)
        self.rentals.pop()

        print("Total Bill:", bill)

    def totalCost(self):
        total = 0

        for bike in self.bikes:
            total += bike.get_price()

        return total


customer1 = Customer("Ahmed", 20)

bike_shop = MainBikeRental()

bike_shop.addBike(Bike("Bike1", 1000))
bike_shop.addBike(Bike("Bike2", 1200))
bike_shop.addBike(Bike("Bike3", 1500))

print("Total Cost:", bike_shop.totalCost())

bike_shop.requestBike(customer1, 2)

print("Available Bikes:", len(bike_shop.get_bikes()))

bike_shop.returnBike(3, 2)

print("Available Bikes:", len(bike_shop.get_bikes()))
print("Total Cost:", bike_shop.totalCost())
```
